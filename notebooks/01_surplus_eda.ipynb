# Surplus exploratory analysis
All records are synthetic and for educational experiments only.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv('../data/synthetic/daily_operations.csv',parse_dates=['date'])
print(df.groupby('category')[['prepared_kg','sold_kg','surplus_kg']].sum())
df.groupby('category').surplus_kg.sum().plot.bar(title='Synthetic surplus by category')
plt.show()
```
